Q10 End-to-End Data Wrangling Task

Perform the following workflow:

Filter the dataset where Sales > 3000

Group data by Region and Product

Calculate total sales

Reshape the result into a pivot table

Explain the business insights obtained from this analysis.


```python
filtered=df[df['Sales']>3000]
```

```python
filtered
```

```python
grouped=filtered.groupby(['Region','Product'])['Sales'].sum()
```

```python
grouped
```

```python
pivot_table = grouped.unstack()
```

```python
pivot_table
```

Business Insights:

1.Which region generates the highest sales.

2.Which products perform best in each region.

3.Which Region + Product combination has the highest sales.
